```python
import re
```

```python
def clean_text(text):
    text = str(text)
    text = re.sub(r"https\S+\www\S+","",text)
    text = re.sub(r"<.*?>","",text)
    text = re.sub(r"\s+"," ",text)
    text = text.strip()

    return text
```

```python
news["text"] = news["text"].apply(clean_text)
news["title"] = news["title"].apply(clean_text)
```

```python
news.head()
```

```python
news["content"] = news["title"] + " " + news["text"]
```

```python
news = news[["content","label"]]
```
